# ¿Qué es Markdown?

Colab tiene dos tipos de celdas: texto y código. Las celdas de texto se formatean usando un lenguaje de marcado simple llamado Markdown.

Para ver el código fuente Markdown, haz doble clic en una celda de texto, mostrando tanto el código como la versión renderizada. Encima del código Markdown hay una barra de herramientas para ayudar en la edición.

## Referencia

| Markdown                    | Vista previa         |
|----------------------------|---------------------|
| `**texto en negrita**`     | **texto en negrita**|
| `*texto en cursiva*` o `_texto en cursiva_` | *texto en cursiva* |
| `` `Monoespaciado` ``      | `Monoespaciado`     |
| `~~tachado~~`              | ~~tachado~~         |
| `[Un enlace](https://www.google.com)` | [Un enlace](https://www.google.com) |
| `![Una imagen](https://www.google.com/images/rss.png)` | ![Una imagen](https://www.google.com/images/rss.png) |
